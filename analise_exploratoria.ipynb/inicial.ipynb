```python
#Passo a passo do projeto
#pip install pandas openpyxl nbformat ipykerkel plotly

#Passo 1 Abrir a bae de dados (importar a base de dados)
import pandas as pd

tabela = pd.read_csv("clientes_treino_churn.csv")
display(tabela)

#Passo 2 Visualizar a base de dados e julgar o que e util/inutil
tabela = tabela.drop(columns= "ID_Cliente" )
display(tabela)
```

```python
#Passo 3: Corrigir o problemas da base de dados(tratamento de dados)
display(tabela.info())
#Nesse caso não temos problemas com a nossa base de dados, então não sera necessarios esse passo. Mas se precisar, usaria o seguinte 

tabela = tabela.dropna()
display(tabela.info())
```

```python
#Passo 4: Analise inicial (entender quantos clientes cancelaram)

#Quantidade - Quantos caras são "0" e quantos são "1" na coluna que cancelou
display(tabela["Cancelou"].value_counts())

#Porcentagem
#display(tabela["Cancelou"].value_counts(normalize=True))
display(tabela["Cancelou"].value_counts(normalize=True).map("{:.1%}".format))
```

```python
#Passo 5: Analise detalhada (causa do cancelamento dos clientes, como cada coluna impacta na cancelamento da assinatura)
import plotly.express as px 
print(tabela.columns)
for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color="Cancelou", text_auto=True)
    grafico.show()

```

```python
#Passo 6: Como resolver os cancelamentos ?

#Podemos analisar que 58.7% dos clientes cancelaram  a assinatura, algo esta errado

#Assinaturas por mês são as mais canceladas - Podemos tentar fazer uma promoçao para incentivar o cliente a mudar o seu plano.
condicao = tabela["Duracao_Contrato"] != "Monthly"
tabela = tabela[condicao]

#Apos 8/9 dias de atraso o indice de cancelamento aumentou
condicao = tabela["Dias_Atraso"] <= 8
tabela = tabela[condicao]

#Apos 2 ligacoes o indice de cancelamentou aumentou, na 9 ligacao o indice de cancelamentou chegou no seu apice 
condicao = tabela["Ligacoes_Callcenter"] <= 2
tabela = tabela[condicao]


display(tabela["Cancelou"].value_counts(normalize=True).map("{:.1%}".format))

#Resolvendo esses 3 problemas saimos de 58% para 26%
```
